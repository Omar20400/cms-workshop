```python
import numpy
import os
```

```python
file_location = os.path.join('data','water.xyz')
```

```python
xyz_file = numpy.genfromtxt(
    fname=file_location,
    skip_header=2,
    dtype='unicode'
)
```

```python
symbols = xyz_file[:,0]
```

```python
coordinates = xyz_file[:,1:]
```

```python
coordinates = coordinates.astype(float)
```

```python
num_atoms = len(symbols)
```

```python
for num1 in range(0, num_atoms):
    print(num1)
```

```python
print(num_atoms)
```

```python
for num1 in range(0, num_atoms):
    print(num1)
```

```python
for num1 in range(0, num_atoms):
    for num2 in range(0, num_atoms):
        print(num1, num2)
```

```python
for num1 in range(0, num_atoms):
    for num2 in range(0, num_atoms):
        print(symbols[num1], "to", symbols[num2])
```

```python
for num1 in range(0, num_atoms):
    for num2 in range(0, num_atoms):
        x_distance = coordinates[num1, 0] - coordinates[num2, 0]
        print(x_distance)
```

```python
for num1 in range(0, num_atoms):
    for num2 in range(0, num_atoms):
        x_distance = coordinates[num1, 0] - coordinates[num2, 0]
        y_distance = coordinates[num1, 1] - coordinates[num2, 1]
        z_distance = coordinates[num1, 2] - coordinates[num2, 2]

        print(x_distance, y_distance, z_distance)
```

```python
for num1 in range(0, num_atoms):
    for num2 in range(0, num_atoms):
        x_distance = coordinates[num1, 0] - coordinates[num2, 0]
        y_distance = coordinates[num1, 1] - coordinates[num2, 1]
        z_distance = coordinates[num1, 2] - coordinates[num2, 2]

        bond_length_12 = numpy.sqrt(
            x_distance**2 +
            y_distance**2 +
            z_distance**2
        )

        print(bond_length_12)
```

```python
for num1 in range(0, num_atoms):
    for num2 in range(0, num_atoms):
        x_distance = coordinates[num1, 0] - coordinates[num2, 0]
        y_distance = coordinates[num1, 1] - coordinates[num2, 1]
        z_distance = coordinates[num1, 2] - coordinates[num2, 2]

        bond_length_12 = numpy.sqrt(
            x_distance**2 +
            y_distance**2 +
            z_distance**2
        )

        print(f"{symbols[num1]} to {symbols[num2]} : {bond_length_12:.3f}")
```

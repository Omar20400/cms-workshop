Plotting and Data Visualization


```python
import os
import numpy
```

```python
distance_file = os.path.join('data', 'distance_data_headers.csv')
```

```python
distances = numpy.genfromtxt(
    fname=distance_file,
    delimiter=',',
    dtype='unicode'
)
```

```python
headers = distances[0]
```

```python
print(headers)
```

```python
data = distances[1:]
```

```python
data = data.astype(float)
```

```python
import matplotlib.pyplot as plt
```

```python
print(data.shape)
```

```python
print(data[:5])
```

```python
plt.plot(data[:, 0], data[:, 1])
```

```python
plt.xlabel('Time')
```

```python
plt.ylabel('Distance')
```

```python
plt.title('Distance vs Time')
```

```python
import matplotlib.pyplot as plt

plt.plot(data[:, 0], data[:, 1])

plt.xlabel('Time')
plt.ylabel('Distance')
plt.title('Time and Distance')

plt.show()
```

```python
plt.plot(data[:, 0], data[:, 1], marker='o')
plt.xlabel('Time')
plt.ylabel('Distance')
plt.title('Time and Distance')
plt.show()
```

```python
plt.plot(data[:, 0], data[:, 1])
plt.plot(data[:, 0], data[:, 2])

plt.xlabel('Time')
plt.ylabel('Distance')
plt.title('Distances vs Time')

plt.show()
```

```python
plt.plot(data[:, 0], data[:, 1], label='Distance 1')
plt.plot(data[:, 0], data[:, 2], label='Distance 2')

plt.xlabel('Time')
plt.ylabel('Distance')
plt.title('Distances vs Time')

plt.legend()

plt.show()
```

```python
plt.plot(data[:, 0], data[:, 1], linestyle='-', label='Distance 1')
plt.plot(data[:, 0], data[:, 2], linestyle='--', label='Distance 2')

plt.xlabel('Time')
plt.ylabel('Distance')
plt.title('Distances vs Time')

plt.legend()

plt.show()
```

```python
plt.plot(data[:, 0], data[:, 1], linestyle='-', label='Distance 1')
plt.plot(data[:, 0], data[:, 2], linestyle='--', label='Distance 2')

plt.xlabel('Time')
plt.ylabel('Distance')
plt.title('Distances vs Time')

plt.legend()

plt.savefig('distance_plot.png')

plt.show()
```

```python
import matplotlib.pyplot as plt

fig = plt.figure()

plt.plot(data[:, 0], data[:, 1])

plt.xlabel('Time')
plt.ylabel('Distance')
plt.title('Distance vs Time')

plt.show()
```

```python
fig = plt.figure(figsize=(8, 5))

plt.plot(data[:, 0], data[:, 1])

plt.xlabel('Time')
plt.ylabel('Distance')
plt.title('Distance vs Time')

plt.show()
```

```python
fig, ax = plt.subplots()
```

```python
fig, ax = plt.subplots()

ax.plot(data[:, 0], data[:, 1])

ax.set_xlabel('Time')
ax.set_ylabel('Distance')
ax.set_title('Distance vs Time')

plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(data[:, 0], data[:, 1])
ax[0].set_xlabel('Time')
ax[0].set_ylabel('Distance 1')
ax[0].set_title('Distance 1 vs Time')

ax[1].plot(data[:, 0], data[:, 2])
ax[1].set_xlabel('Time')
ax[1].set_ylabel('Distance 2')
ax[1].set_title('Distance 2 vs Time')

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(data[:, 0], data[:, 1])
ax[0].set_xlabel('Time')
ax[0].set_ylabel('Distance 1')
ax[0].set_title('Distance 1 vs Time')

ax[1].plot(data[:, 0], data[:, 2])
ax[1].set_xlabel('Time')
ax[1].set_ylabel('Distance 2')
ax[1].set_title('Distance 2 vs Time')

plt.tight_layout()

fig.savefig('distance_subplots.png')

plt.show()
```

```python
import os

datafile = open("energies.txt", "w+")

for f in filenames:

    # استخراج اسم الجزيء
    file_name = os.path.basename(f)
    split_filename = file_name.split(".")
    molecule_name = split_filename[0]

    # قراءة الملف
    outfile = open(f, "r")
    data = outfile.readlines()
    outfile.close()

    # البحث عن الطاقة
    for line in data:
        if "Final Energy" in line:
            words = line.split()
            energy = float(words[3])

            # كتابة النتيجة في الملف
            datafile.write(f"{molecule_name}\t{energy}\n")

datafile.close()
```

```python
datafile = open("energies.txt", "w+")

for f in filenames:

    file_name = os.path.basename(f)
    molecule_name = file_name.split('.')[0]

    outfile = open(f, "r")
    data = outfile.readlines()
    outfile.close()

    for line in data:
        if "Final Energy" in line:
            words = line.split()
            energy = float(words[3])

            datafile.write(f"For the file {molecule_name} the energy is {energy} in kcal/mole.\n")

datafile.close()
```
